¿Qué es un Algoritmo y cuáles son sus características clave?
Un algoritmo se define de manera general como un método sistemático, constituido por una serie o secuencia finita de pasos o instrucciones precisas y no ambiguas, diseñado para resolver un problema o calcular una función
. Formalmente, se le describe como un "conjunto de pasos que, ejecutados de la manera correcta, permiten obtener un resultado en un tiempo acotado"
.
Sus características clave para que se considere como tal son
:
Entrada y Salida: Debe tener definidos de forma clara los datos que requiere para operar (entradas) y el resultado específico que produce tras procesarlos (salidas)
.
Precisión: Debe indicar sin ambigüedades el orden exacto de los pasos a seguir y la manera precisa en la que estos se realizan
.
Finitud: El algoritmo siempre debe terminar después de concluir un número finito de pasos
.
Determinismo: Para un mismo conjunto de datos proporcionado como entrada, el algoritmo debe transitar por los mismos estados y producir siempre exactamente el mismo resultado
.
Efectividad: Cada instrucción debe ser lo suficientemente básica como para poder completarse en un tiempo finito y lograr el efecto esperado
.
Generalidad: Debe ser lo suficientemente general como para contemplar todos los casos válidos de datos de entrada dentro del problema planteado
.
El Problema del Paro (Halting Problem) de Alan Turing
Formulado por el matemático Alan Turing en 1936, el Problema del Paro es un pilar fundamental en la teoría de la computabilidad
. Consiste en determinar si es posible diseñar un algoritmo general que, al recibir como entrada cualquier programa de computadora y un conjunto de datos para este, sea capaz de decidir correctamente si dicho programa finalizará en algún momento (se detendrá) o si continuará ejecutándose indefinidamente (entrará en un ciclo infinito)
.
Turing demostró matemáticamente que este problema es indecidible
. Esto significa que es imposible que exista un algoritmo o programa de computadora universal capaz de resolver de manera correcta todas las instancias de este problema
. Intentar resolverlo simplemente corriendo el programa y esperando a ver si termina no es un algoritmo válido, ya que si el programa entra en un ciclo infinito, nos quedaríamos esperando eternamente sin poder determinar con certeza si se detendrá en el paso siguiente o si nunca lo hará
.

2. Utiliza la librería `time` o la función mágica `%timeit` para comparar el tiempo consumido (A posteriori) entre el Fibonacci recursivo y el iterativo para `n=35`.

````python
import time

# Fibonacci recursivo
def fibonacci_recursivo(n):
    if n <= 1:
        return n
    return fibonacci_recursivo(n - 1) + fibonacci_recursivo(n - 2)


# Fibonacci iterativo
def fibonacci_iterativo(n):
    if n <= 1:
        return n

    lofib = 0
    hifib = 1

    for i in range(2, n + 1):
        temporal = lofib + hifib
        lofib = hifib
        hifib = temporal

    return hifib


n = 35

# Medir Fibonacci recursivo
inicio = time.time()
resultado_recursivo = fibonacci_recursivo(n)
fin = time.time()

tiempo_recursivo = fin - inicio


# Medir Fibonacci iterativo
inicio = time.time()
resultado_iterativo = fibonacci_iterativo(n)
fin = time.time()

tiempo_iterativo = fin - inicio


print("Fibonacci de", n, ":", resultado_recursivo)
print("Tiempo recursivo:", tiempo_recursivo, "segundos")
print("Fibonacci de", n, ":", resultado_iterativo)
print("Tiempo iterativo:", tiempo_iterativo, "segundos")
````
